```python
#IMportamos librerias necesarias:
import os
import pandas  as pd
from datetime import datetime
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
import math as mt
from scipy.stats import pearsonr

sns.set_style('darkgrid')
warnings.filterwarnings('ignore')
```

```python
# Creamos dataframe con los datos
df = pd.read_excel('datosejercicioevaluacionanchuras.xlsx')
# print(df)
```

```python
#Separamos datos en sub conjuntos predinástico temprano y predinástico tardío
df_1 = df[df['Época histórica']==1]
df_2 = df[df['Época histórica']==2]
# print(df_2)
```

```python
# Calculamos los distintos valores para el predinástico temprano
valores_temprano = pd.DataFrame(df_1['Anchura del cráneo'].describe())
valores_temprano.reset_index(inplace=True)
valores_temprano.rename(columns={'index':'Medida'},inplace = True)
# print(valores_temprano)
moda_temprano  = df_1['Anchura del cráneo'].mode()
# print(moda_temprano) Se tienen dos modas
rango_temprano = df_1['Anchura del cráneo'].max() - df_1['Anchura del cráneo'].min()
varianza_temprano = df_1['Anchura del cráneo'].var()
media_temprano = valores_temprano.loc[valores_temprano['Medida']=='mean', 'Anchura del cráneo'].values[0]
mediana_temprano = valores_temprano.loc[valores_temprano['Medida']=='50%', 'Anchura del cráneo'].values[0]
desviacion_temprano = valores_temprano.loc[valores_temprano['Medida']=='std', 'Anchura del cráneo'].values[0]

#Calculanos XI - media
df_1['Xi - Media'] = df_1['Anchura del cráneo'] - media_temprano
df_1['(Xi - Media)*3'] = df_1['Xi - Media']**3
df_1['(Xi - Media)*4'] = df_1['Xi - Media']**4

#Coeficiente de variación de pearson
coef_var_P_temprano = desviacion_temprano/media_temprano
#Coeficiente de fisher
Coef_Fisher_temprano =df_1['(Xi - Media)*3'].sum()/((df_1['Época histórica'].count()-1)*mt.pow(desviacion_temprano,3))
#Coeficiente de curtosis
coeficiente_Curtosis_temprano = df_1['(Xi - Media)*4'].sum()/((df_1['Época histórica'].count()-1)*mt.pow(desviacion_temprano,4)) -3

#IMprimimos para ver si ha ido bien:

print("Valores varios: ",valores_temprano, "\n")
print("Moda: ",moda_temprano, "\n")
print("rango: ",rango_temprano, "\n")
print("varianza: ",varianza_temprano, "\n")
print("media: ",media_temprano, "\n")
print("mediana: ",mediana_temprano, "\n")
print("desviacion: ",desviacion_temprano, "\n")
print("Variacion de pearson: ",coef_var_P_temprano, "\n")
print("Fisher: ",Coef_Fisher_temprano, "\n")
print("Curtosis: ",coeficiente_Curtosis_temprano, "\n")
```

```python
# Calculamos los distintos valores para el predinástico tardio
valores_tardio = pd.DataFrame(df_2['Anchura del cráneo'].describe())
valores_tardio.reset_index(inplace=True)
valores_tardio.rename(columns={'index':'Medida'},inplace = True)
# print(valores_tardio)
moda_tardio  = df_2['Anchura del cráneo'].mode()
# print(moda_tardio) Se tienen dos modas
rango_tardio = df_2['Anchura del cráneo'].max() - df_2['Anchura del cráneo'].min()
varianza_tardio = df_2['Anchura del cráneo'].var()
media_tardio = valores_tardio.loc[valores_tardio['Medida']=='mean', 'Anchura del cráneo'].values[0]
mediana_tardio = valores_tardio.loc[valores_tardio['Medida']=='50%', 'Anchura del cráneo'].values[0]
desviacion_tardio = valores_tardio.loc[valores_tardio['Medida']=='std', 'Anchura del cráneo'].values[0]

#Calculanos XI - media
df_2['Xi - Media'] = df_2['Anchura del cráneo'] - media_tardio
df_2['(Xi - Media)*3'] = df_2['Xi - Media']**3
df_2['(Xi - Media)*4'] = df_2['Xi - Media']**4

#Coeficiente de variación de pearson
coef_var_P_tardio = desviacion_tardio/media_tardio
#Coeficiente de fisher
Coef_Fisher_tardio =df_2['(Xi - Media)*3'].sum()/((df_2['Época histórica'].count()-1)*mt.pow(desviacion_tardio,3))
#Coeficiente de curtosis
coeficiente_Curtosis_tardio = df_2['(Xi - Media)*4'].sum()/((df_2['Época histórica'].count()-1)*mt.pow(desviacion_tardio,4)) -3

#IMprimimos para ver si ha ido bien:
print("Valores varios: ",valores_tardio, "\n")
print("Moda: ",moda_tardio, "\n")
print("rango: ",rango_tardio, "\n")
print("varianza: ",varianza_tardio, "\n")
print("media: ",media_tardio, "\n")
print("mediana: ",mediana_tardio, "\n")
print("desviacion: ",desviacion_tardio, "\n")
print("Variacion de pearson: ",coef_var_P_tardio, "\n")
print("Fisher: ",Coef_Fisher_tardio, "\n")
print("Curtosis: ",coeficiente_Curtosis_tardio, "\n")
```

```python
# !pip install scikit-learn
# Ahora vamos al apartado B para determinar si cada una de las dos sub muestras siguen una 
# distribución normal utilizando el test  de Kolmogorov-Smirnov. 
from scipy import stats
import numpy as np
from sklearn.preprocessing import StandardScaler

# conversion de los df filtrado a np.array para aprovechar sus caracteristicas
submuestra1 = np.array(df_1['Anchura del cráneo'])
submuestra2 = np.array(df_2['Anchura del cráneo'])

# Asegurarse de que las submuestras sean arreglos de una sola columna
submuestra1 = submuestra1.reshape(-1, 1)
submuestra2 = submuestra2.reshape(-1, 1)

# Inicializar standarscaler de scktlearn
scaler1 = StandardScaler()
scaler2 = StandardScaler()

# Normalizar las muestras
submuestra1 = scaler1.fit_transform(submuestra1)
submuestra2 = scaler2.fit_transform(submuestra2)

#llevar las submuestras a matriz unidmensional
submuestra1=submuestra1.flatten()
submuestra2=submuestra2.flatten()

# Aplicar la prueba de Kolmogorov-Smirnov
statistic1, pvalue1 = stats.kstest(submuestra1, 'norm')
statistic2, pvalue2 = stats.kstest(submuestra2, 'norm')

# Imprimir los resultados
print("Resultados de la submuestra 1:")
print(f"Estadístico: {statistic1}, Valor p: {pvalue1}")
if pvalue1 > 0.05:
    print("La submuestra 1 sigue una distribución normal")
else:
    print("La submuestra 1 no sigue una distribución normal")

print("\nResultados de la submuestra 2:")
print(f"Estadístico: {statistic2}, Valor p: {pvalue2}")
if pvalue2 > 0.05:
    print("La submuestra 2 sigue una distribución normal")
else:
    print("La submuestra 2 no sigue una distribución normal")
```

```python
from scipy.stats import f
## Calculamos grados libertad
grados_libertad_temprano = valores_temprano.loc[valores_temprano['Medida']=='count', 'Anchura del cráneo'].values[0] - 1
grados_libertad_tardio = valores_tardio.loc[valores_tardio['Medida']=='count', 'Anchura del cráneo'].values[0] - 1

# Calculamos alfa para 90% de confianza (al ser 2 colas -> alfa/2)
alfa_doscolas = 0.1/2

# calculamos valor crítico inferior y superior
critico_superior = f.ppf(1-alfa_doscolas,grados_libertad_temprano,grados_libertad_tardio)
critico_inferior = 1/critico_superior

# Calulamos la F estadística

f_estadistica = mt.pow (valores_temprano.loc[valores_temprano['Medida'] == 'std', 'Anchura del cráneo'].values[0],2)/mt.pow(valores_tardio.loc[valores_tardio['Medida'] == 'std', 'Anchura del cráneo'].values[0],2)

print("Grados libertad temprano: " , grados_libertad_temprano, "\n")
print("Grados libertad tardio: " , grados_libertad_tardio, "\n")
print("Alfa dos colas: " , alfa_doscolas, "\n")
print("Critico superior e inferior: " , critico_inferior, " < F <" ,critico_superior , "\n")
print("F estadistica: ", f_estadistica)
```

```python
# Creamos gráfico de demostración:
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import f

f_values = np.linspace(0.01, 5, 1000)

f_distribucion = f.pdf(f_values, grados_libertad_temprano, grados_libertad_tardio)

plt.plot(f_values, f_distribucion, label='Distribución F')

plt.vlines(critico_inferior, 0, f.pdf(critico_inferior, grados_libertad_temprano, grados_libertad_tardio), colors='r', label='Valor Crítico Inferior')
plt.vlines(critico_superior, 0, f.pdf(critico_superior, grados_libertad_temprano, grados_libertad_tardio), colors='r', label='Valor Crítico Superior')
plt.vlines(f_estadistica, 0, f.pdf(f_estadistica, grados_libertad_temprano, grados_libertad_tardio), colors='g', label='Valor Observado')

plt.legend()
plt.xlabel('Valor F')
plt.ylabel('Densidad de Probabilidad')
plt.title('Distribución F con valores críticos y valor observado')

plt.show()
```

```python
# Calculamos el valor critico T y los intervalos de confianza para 90, 95 y 99
from scipy.stats import t

diferencia_medias = valores_temprano.loc[valores_temprano['Medida'] == 'mean', 'Anchura del cráneo'].values[0] - valores_tardio.loc[valores_tardio['Medida'] == 'mean', 'Anchura del cráneo'].values[0]
n1=valores_temprano.loc[valores_temprano['Medida'] == 'count', 'Anchura del cráneo'].values[0]
n2=valores_tardio.loc[valores_tardio['Medida'] == 'count', 'Anchura del cráneo'].values[0]
s1 = valores_temprano.loc[valores_temprano['Medida'] == 'std', 'Anchura del cráneo'].values[0]
s2 = valores_tardio.loc[valores_tardio['Medida'] == 'std', 'Anchura del cráneo'].values[0]
grados_libertad = n1+n2-2
error_standar  = mt.sqrt((((n1*mt.pow(s1,2)+n2*mt.pow(s2,2))*((1/n1)+(1/n2)))/(grados_libertad)))
#el error standar para la diferencia de medias, es el determinado ya que las varianzas poblacionales son desconocidas pero iguales

intervalo_90 = stats.t.interval(0.90,grados_libertad,loc=diferencia_medias,scale=error_standar)
intervalo_95 = stats.t.interval(0.95,grados_libertad,loc=diferencia_medias,scale=error_standar)
intervalo_99 = stats.t.interval(0.99,grados_libertad,loc=diferencia_medias,scale=error_standar)

print("los intervalos de confianza para la diferencia de medias con un 90% de confianza es:" ,intervalo_90)
print("--------------------------------------------------------------------------------------------------------------")
print("los intervalos de confianza para la diferencia de medias con un 95% de confianza es:" ,intervalo_95)
print("--------------------------------------------------------------------------------------------------------------")
print("los intervalos de confianza para la diferencia de medias con un 99% de confianza es:" ,intervalo_99)
```

```python
#calculo de valores para el test T
alfa = 0.05
alfa_medios = alfa/2

lim_sup = t.ppf(1-alfa_medios,grados_libertad)
lim_inf = t.ppf(alfa_medios,grados_libertad)

print('grados de libertad: ', grados_libertad, '\n')
print('alfa medios: ', alfa_medios, '\n')
print("la zona de aceptacion para la prueba es :",lim_inf,",",lim_sup)
```

```python
# calculamos el valor t estadistico:
valor_critico_t =  float(diferencia_medias)/(error_standar)

print("el valor critico de T es :", t)
```
